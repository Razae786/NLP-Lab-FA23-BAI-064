## Task 1: Multi-Page PDF Extraction with Word Frequency Analysis

- **(a)** Open a PDF file using `PyPDF2` or `pdfplumber` and extract text from each page individually. Store the result as a list of strings where each index represents one page. Print the total number of pages extracted.
- **(b)** Merge all page strings into a single string. Use `re.sub()` to remove all punctuation and digits, convert to lowercase, and split on whitespace to produce a token list. Print the total token count.
- **(c)** Build a word frequency dictionary from the token list. Remove at least 20 common stopwords. Print the top 10 most frequent words along with their counts in descending order.
- **(d)** Accept a keyword from the user via `input()`. Search through the page-wise list, split each page's text into sentences, and print every sentence containing the keyword along with its page number. If no match is found, print an appropriate message.


```python
import re
import PyPDF2

# (a) Open PDF and extract text from each page individually
pdf_path = "sample_nlp.pdf"
pages_text = []
with open(pdf_path, "rb") as f:
    reader = PyPDF2.PdfReader(f)
    for page in reader.pages:
        pages_text.append(page.extract_text() or "")

print(f"(a) Total pages extracted: {len(pages_text)}")

# (b) Merge all page strings into a single string, remove punctuation and digits, lowercase, and tokenize
merged_text = " ".join(pages_text)
cleaned_text = re.sub(r"[^\w\s]|[\d_]", "", merged_text).lower()
tokens = cleaned_text.split()
print(f"(b) Total token count: {len(tokens)}")

# (c) Word frequency dictionary removing at least 20 common stopwords
stopwords_20 = {
    "the", "and", "is", "in", "to", "of", "for", "a", "with", "that",
    "are", "on", "as", "by", "this", "it", "an", "be", "or", "from",
    "can", "we", "will", "at", "more", "their", "there", "which"
}
freq = {}
for token in tokens:
    if token not in stopwords_20:
        freq[token] = freq.get(token, 0) + 1

sorted_freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)
print("\n(c) Top 10 most frequent words (excluding stopwords):")
for word, count in sorted_freq[:10]:
    print(f"    {word:<15}: {count}")

# (d) Search keyword across page-wise list, split into sentences, and print matches with page number
try:
    keyword = input("\nEnter keyword to search (default: 'data'): ").strip().lower() or "data"
except (EOFError, OSError):
    keyword = "data"

matches = []
for page_num, page_str in enumerate(pages_text, start=1):
    sentences = [s.strip() for s in re.split(r"[.!?]", page_str) if s.strip()]
    for sent in sentences:
        if keyword in sent.lower():
            matches.append((page_num, sent))

print(f"\n(d) Searching for keyword: '{keyword}'")
if matches:
    print(f"Found {len(matches)} matching sentence(s):")
    for p_num, sent in matches:
        print(f"  [Page {p_num}] {sent}")
else:
    print(f"No sentences found containing the keyword '{keyword}'.")
```

## Task 2: Structured Page Metadata Extraction and CSV Export

- **(a)** Open a multi-page PDF (minimum 5 pages). For each page, build a dictionary with exactly these keys: `page_number`, `raw_text`, `word_count`, and `char_count` (excluding spaces). Collect all dictionaries into a list called `pages_data` and print the first 2 entries.
- **(b)** From `pages_data`, identify and print: (i) the page with the highest word count and its full content, (ii) the page with the lowest non-zero word count, and (iii) the average word count across all pages rounded to 2 decimal places. Handle pages that return zero words.
- **(c)** Accept a search phrase via `input()`. Perform a case-insensitive search across all pages. For every page containing the phrase, print its page number and how many times the phrase appears. If no page contains the phrase, raise a custom exception named `PhraseNotFoundError` and catch it gracefully.
- **(d)** Using Python's built-in `csv` module (not pandas), write a file named `pdf_summary.csv` with columns: `page_number`, `word_count`, `char_count`. Then re-read that file and print its contents row by row to confirm the write was successful.


```python
import csv
import pprint
import PyPDF2

# (a) Open multi-page PDF (>= 5 pages) and build metadata dictionary for each page
pages_data = []
with open("sample_nlp.pdf", "rb") as f:
    reader = PyPDF2.PdfReader(f)
    for i, page in enumerate(reader.pages, start=1):
        raw = page.extract_text() or ""
        words = raw.split()
        chars_no_spaces = len(raw.replace(" ", "").replace("\n", "").replace("\r", "").replace("\t", ""))
        pages_data.append({
            "page_number": i,
            "raw_text": raw,
            "word_count": len(words),
            "char_count": chars_no_spaces
        })

print("(a) First 2 entries in pages_data:")
for entry in pages_data[:2]:
    preview = dict(entry)
    preview["raw_text"] = preview["raw_text"][:90].replace("\n", " ") + "..."
    pprint.pprint(preview)

# (b) Statistical analysis of word counts
page_max = max(pages_data, key=lambda p: p["word_count"])
print(f"\n(b)(i) Page with highest word count: Page {page_max['page_number']} ({page_max['word_count']} words)")
print("Full content:\n" + page_max["raw_text"].strip())

non_zero_pages = [p for p in pages_data if p["word_count"] > 0]
if non_zero_pages:
    page_min = min(non_zero_pages, key=lambda p: p["word_count"])
    print(f"\n(b)(ii) Page with lowest non-zero word count: Page {page_min['page_number']} ({page_min['word_count']} words)")
else:
    print("\n(b)(ii) No pages with non-zero words found.")

total_words = sum(p["word_count"] for p in pages_data)
avg_words = round(total_words / len(pages_data), 2) if pages_data else 0.0
print(f"(b)(iii) Average word count across all pages: {avg_words}")

# (c) Custom exception and case-insensitive phrase search
class PhraseNotFoundError(Exception):
    pass

def search_phrase_in_pages(phrase, pages):
    phrase_lower = phrase.lower()
    matches = 0
    for p in pages:
        count = p["raw_text"].lower().count(phrase_lower)
        if count > 0:
            matches += 1
            print(f"  Page {p['page_number']}: appears {count} time(s)")
    if matches == 0:
        raise PhraseNotFoundError(f"The phrase '{phrase}' was not found in any page.")

try:
    search_phrase = input("\nEnter search phrase (default: 'data acquisition'): ").strip() or "data acquisition"
except (EOFError, OSError):
    search_phrase = "data acquisition"

print(f"\n(c) Searching for phrase: '{search_phrase}'")
try:
    search_phrase_in_pages(search_phrase, pages_data)
except PhraseNotFoundError as err:
    print(f"Caught Expected Exception: {err}")

# Demonstrate raising and catching PhraseNotFoundError with an absent phrase
print("\nDemonstrating PhraseNotFoundError with absent phrase ('quantum physics'):")
try:
    search_phrase_in_pages("quantum physics", pages_data)
except PhraseNotFoundError as err:
    print(f"Caught Expected Exception: {err}")

# (d) Export summary to CSV using built-in csv module and verify
csv_filename = "pdf_summary.csv"
with open(csv_filename, mode="w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["page_number", "word_count", "char_count"])
    for p in pages_data:
        writer.writerow([p["page_number"], p["word_count"], p["char_count"]])

print(f"\n(d) Re-reading '{csv_filename}' row by row:")
with open(csv_filename, mode="r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(f"  {row}")
```

## Task 3: PDF Text Preprocessing Pipeline for NLP

- **(a)** Open any PDF file and extract all text into one string. Apply the following cleaning steps in order using `re.sub()`: (i) remove all URLs, (ii) remove all digits, (iii) remove all non-alphabetic characters except spaces and full stops, (iv) collapse multiple consecutive spaces into a single space. Print the character count before and after cleaning.
- **(b)** Convert the cleaned text to lowercase. Split into a list of sentences using `re.split(r'[.!?]', text)`. Strip each sentence of leading and trailing whitespace. Remove empty strings. Print the sentence count before and after removing empty strings.
- **(c)** Tokenize each sentence into words. Remove English stopwords using `nltk.corpus.stopwords.words('english')`. Discard any sentence that has fewer than 5 tokens remaining after stopword removal. Print the sentence count before and after this filter and display 3 sample sentences.
- **(d)** Write all remaining sentences to a file named `nlp_ready_output.txt`, one sentence per line. Re-read the file and print: (i) total lines written, (ii) the longest sentence by character length, and (iii) the shortest sentence by character length.


```python
import re
import PyPDF2
import nltk
from nltk.corpus import stopwords

# (a) Extract text and apply 4 cleaning steps in sequence using re.sub()
with open("sample_nlp.pdf", "rb") as f:
    reader = PyPDF2.PdfReader(f)
    raw_all = " ".join(page.extract_text() or "" for page in reader.pages)

print(f"(a) Character count before cleaning: {len(raw_all)}")

# (i) remove URLs
clean1 = re.sub(r"https?://\S+|www\.\S+", "", raw_all)
# (ii) remove all digits
clean2 = re.sub(r"\d+", "", clean1)
# (iii) remove all non-alphabetic characters except spaces and full stops
clean3 = re.sub(r"[^a-zA-Z\s.]", "", clean2)
# (iv) collapse multiple consecutive spaces into a single space
clean4 = re.sub(r"\s+", " ", clean3).strip()

print(f"(a) Character count after cleaning: {len(clean4)}")

# (b) Lowercase, split into sentences, strip whitespace, remove empty strings
cleaned_lower = clean4.lower()
sentences_raw = [s.strip() for s in re.split(r"[.!?]", cleaned_lower)]
print(f"\n(b) Sentence count before removing empty strings: {len(sentences_raw)}")

sentences = [s for s in sentences_raw if s]
print(f"(b) Sentence count after removing empty strings: {len(sentences)}")

# (c) Tokenize words, remove stopwords, and filter out sentences with < 5 non-stopwords
nltk.download('stopwords', quiet=True)
stop_words = set(stopwords.words("english"))

filtered_sentences = []
for sent in sentences:
    tokens = [w for w in sent.split() if w not in stop_words]
    if len(tokens) >= 5:
        filtered_sentences.append(sent)

print(f"\n(c) Sentence count before filter: {len(sentences)}")
print(f"(c) Sentence count after filter (>= 5 tokens): {len(filtered_sentences)}")
print("3 Sample Sentences:")
for s in filtered_sentences[:3]:
    print(f"  - {s}")

# (d) Write to nlp_ready_output.txt, re-read, and display statistics
output_file = "nlp_ready_output.txt"
with open(output_file, "w", encoding="utf-8") as f:
    for s in filtered_sentences:
        f.write(s + "\n")

with open(output_file, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f if line.strip()]

print(f"\n(d)(i) Total lines written: {len(lines)}")
longest_sentence = max(lines, key=len)
shortest_sentence = min(lines, key=len)
print(f"(d)(ii) Longest sentence ({len(longest_sentence)} chars):\n  {longest_sentence}")
print(f"(d)(iii) Shortest sentence ({len(shortest_sentence)} chars):\n  {shortest_sentence}")
```

## Task 4: DOCX Paragraph Classification and Statistical Analysis

- **(a)** Open any `.docx` file using `python-docx`. For each paragraph, store its text and `style.name` in a list of dictionaries. Print the total paragraph count and all unique style names found in the document.
- **(b)** Classify all paragraphs into exactly three groups: (i) Headings — style name starts with "Heading", (ii) Body — style is "Normal" or contains "Body", (iii) Other/Empty — everything else including blank paragraphs. Print the count of each group.
- **(c)** For every Body paragraph with more than 5 words, compute its word count and character count excluding spaces. Print: (i) the paragraph with the highest word count, (ii) the paragraph with the lowest word count above 5 words, and (iii) the mean word count across all qualifying body paragraphs rounded to 2 decimal places.
- **(d)** Build a nested dictionary structured as `{ style_name: { "count": N, "texts": [ list ] } }` for all styles found. Sort the outer dictionary by count in descending order. If a style has more than 2 paragraphs, show only the first 2 texts. Pretty-print the result.


```python
import docx
import pprint

# (a) Open .docx and store text and style.name in a list of dictionaries
doc = docx.Document("doc_a.docx")
para_records = [{"text": p.text, "style_name": p.style.name} for p in doc.paragraphs]
print(f"(a) Total paragraph count: {len(para_records)}")
unique_styles = sorted(list(set(p["style_name"] for p in para_records)))
print(f"(a) Unique style names: {unique_styles}")

# (b) Classify into exactly three groups: Headings, Body, Other/Empty
headings = []
body = []
other_empty = []

for p in para_records:
    txt = p["text"].strip()
    style = p["style_name"]
    if not txt:
        other_empty.append(p)
    elif style.startswith("Heading"):
        headings.append(p)
    elif style == "Normal" or "Body" in style:
        body.append(p)
    else:
        other_empty.append(p)

print(f"\n(b) Paragraph Classification Counts:")
print(f"    Headings    : {len(headings)}")
print(f"    Body        : {len(body)}")
print(f"    Other/Empty : {len(other_empty)}")

# (c) Body paragraphs with > 5 words: compute word count and char count (excluding spaces)
qualifying_body = []
for p in body:
    words = p["text"].split()
    if len(words) > 5:
        char_count = len(p["text"].replace(" ", "").replace("\t", "").replace("\n", ""))
        qualifying_body.append({
            "text": p["text"],
            "word_count": len(words),
            "char_count": char_count
        })

if qualifying_body:
    highest_p = max(qualifying_body, key=lambda x: x["word_count"])
    lowest_p = min(qualifying_body, key=lambda x: x["word_count"])
    mean_wc = round(sum(x["word_count"] for x in qualifying_body) / len(qualifying_body), 2)
    print(f"\n(c)(i) Highest word count ({highest_p['word_count']} words):\n  {highest_p['text']}")
    print(f"(c)(ii) Lowest word count above 5 words ({lowest_p['word_count']} words):\n  {lowest_p['text']}")
    print(f"(c)(iii) Mean word count: {mean_wc}")

# (d) Nested dictionary sorted by count descending (max 2 sample texts per style)
style_summary = {}
for p in para_records:
    s = p["style_name"]
    if s not in style_summary:
        style_summary[s] = {"count": 0, "texts": []}
    style_summary[s]["count"] += 1
    style_summary[s]["texts"].append(p["text"])

sorted_style_summary = dict(
    sorted(
        [(s, {"count": d["count"], "texts": d["texts"][:2]}) for s, d in style_summary.items()],
        key=lambda item: item[1]["count"],
        reverse=True
    )
)
print("\n(d) Style Summary (Sorted descending, max 2 texts):")
pprint.pprint(sorted_style_summary)
```

## Task 5: DOCX Table Extraction and Pandas Integration

- **(a)** Open a `.docx` file that contains at least one table. Iterate through `doc.tables` and extract each table as a list of lists: `[[row1col1, row1col2, ...], [row2col1, ...]]`. Print the total number of tables found and the dimensions (rows × columns) of each.
- **(b)** Take the first table's data, treat the first row as the header and the remaining rows as data, and create a pandas DataFrame. Print the full DataFrame. Then count and print the number of empty or None cells per column.
- **(c)** Extract all paragraphs that appear before the first table in the document by iterating through `doc.element.body` children and stopping once a `w:tbl` element is encountered. Store these paragraphs in a list and print their count.
- **(d)** Write both the pre-table paragraphs and the first table's data into a single file named `docx_extracted.txt`. The file must have clearly labeled sections: a `--- PRE-TABLE PARAGRAPHS ---` header followed by the paragraphs, then a `--- TABLE DATA ---` header followed by rows formatted as pipe-separated values, e.g., `col1 | col2 | col3`.


```python
import docx
import pandas as pd
from docx.text.paragraph import Paragraph

doc = docx.Document("doc_a.docx")

# (a) Iterate through doc.tables and extract as list of lists
tables_data = []
for t in doc.tables:
    table_matrix = [[cell.text.strip() for cell in row.cells] for row in t.rows]
    tables_data.append(table_matrix)

print(f"(a) Total tables found: {len(tables_data)}")
for idx, tbl in enumerate(tables_data, start=1):
    rows = len(tbl)
    cols = len(tbl[0]) if rows > 0 else 0
    print(f"    Table {idx}: {rows} rows x {cols} columns")

# (b) Create DataFrame from first table: row 0 as header, remaining as rows
header = tables_data[0][0]
rows = tables_data[0][1:]
df_table = pd.DataFrame(rows, columns=header)
print("\n(b) First Table as DataFrame:")
print(df_table)

print("\n(b) Empty or None cells per column:")
empty_cells = (df_table.isna() | (df_table == "")).sum()
print(empty_cells)

# (c) Extract paragraphs before the first table by iterating through doc.element.body children
pre_table_paragraphs = []
for child in doc.element.body:
    if child.tag.endswith("tbl"):
        break
    if child.tag.endswith("p"):
        p_obj = Paragraph(child, doc)
        if p_obj.text.strip():
            pre_table_paragraphs.append(p_obj.text.strip())

print(f"\n(c) Pre-table paragraphs count: {len(pre_table_paragraphs)}")

# (d) Write to docx_extracted.txt with labeled sections and pipe-separated rows
with open("docx_extracted.txt", "w", encoding="utf-8") as f:
    f.write("--- PRE-TABLE PARAGRAPHS ---\n")
    for p in pre_table_paragraphs:
        f.write(p + "\n")
    f.write("\n--- TABLE DATA ---\n")
    for row in tables_data[0]:
        f.write(" | ".join(row) + "\n")

print("\n(d) Successfully written to docx_extracted.txt. File content preview:")
with open("docx_extracted.txt", "r", encoding="utf-8") as f:
    print(f.read().strip())
```

- **(c)** Tokenize every paragraph in the deduplicated list into words (lowercase, punctuation removed via `re.sub()`). Build a combined word frequency dictionary using `collections.Counter`. Remove at least 15 stopwords. Display the top 20 most frequent words.
- **(d)** Compute and print a summary containing: (i) total headings across both documents, (ii) total unique body paragraphs after deduplication, (iii) total unique word types (vocabulary size), (iv) average paragraph length in words rounded to 2 decimal places, and (v) the single longest paragraph by word count from the combined set.


```python
import docx
import re
from collections import Counter

# (a) Open Document A and Document B, extract headings, and perform set operations
doc_a = docx.Document("doc_a.docx")
doc_b = docx.Document("doc_b.docx")

headings_a = [p.text.strip() for p in doc_a.paragraphs if p.style.name.startswith("Heading") and p.text.strip()]
headings_b = [p.text.strip() for p in doc_b.paragraphs if p.style.name.startswith("Heading") and p.text.strip()]

set_a = set(headings_a)
set_b = set(headings_b)

print(f"(a)(i) Common headings: {set_a & set_b}")
print(f"(a)(ii) Headings only in Document A: {set_a - set_b}")
print(f"(a)(iii) Headings only in Document B: {set_b - set_a}")

# (b) Extract Normal paragraphs, merge, and deduplicate
normal_a = [p.text.strip() for p in doc_a.paragraphs if p.style.name == "Normal" and p.text.strip()]
normal_b = [p.text.strip() for p in doc_b.paragraphs if p.style.name == "Normal" and p.text.strip()]

merged_normal = normal_a + normal_b
print(f"\n(b) Total Normal paragraphs before deduplication: {len(merged_normal)}")

seen = set()
deduped_normal = []
for p in merged_normal:
    key = p.lower()
    if key not in seen:
        seen.add(key)
        deduped_normal.append(p)

print(f"(b) Total Normal paragraphs after deduplication: {len(deduped_normal)}")

# (c) Tokenize into words, remove at least 15 stopwords, and build Counter
stopwords_15 = {
    "the", "and", "is", "in", "to", "of", "for", "a", "with", "that",
    "are", "on", "as", "by", "this", "it", "an", "be", "or", "from"
}

all_words = []
for p in deduped_normal:
    clean_p = re.sub(r"[^\w\s]|[\d_]", "", p).lower()
    all_words.extend(clean_p.split())

filtered_words = [w for w in all_words if w not in stopwords_15]
word_counter = Counter(filtered_words)

print("\n(c) Top 20 most frequent words:")
for word, count in word_counter.most_common(20):
    print(f"    {word:<15}: {count}")

# (d) Statistical summary
total_headings = len(headings_a) + len(headings_b)
unique_body_paras = len(deduped_normal)
vocab_size = len(word_counter)
para_lengths = [len(p.split()) for p in deduped_normal]
avg_para_len = round(sum(para_lengths) / len(para_lengths), 2) if para_lengths else 0.0
longest_para = max(deduped_normal, key=lambda p: len(p.split())) if deduped_normal else ""

print(f"\n(d) Summary Statistics:")
print(f"    (i) Total headings across both documents: {total_headings}")
print(f"    (ii) Total unique body paragraphs after deduplication: {unique_body_paras}")
print(f"    (iii) Total unique word types (vocabulary size): {vocab_size}")
print(f"    (iv) Average paragraph length: {avg_para_len} words")
print(f"    (v) Longest paragraph ({len(longest_para.split())} words):\n        {longest_para}")
```

## Task 7: Multi-Endpoint API Data Joining and Analysis

- **(a)** Send GET requests to both endpoints below. For each, verify the HTTP status code is exactly 200; if not, raise a `ConnectionError` with the actual status code in the message. Parse both responses as JSON lists.
  - `https://jsonplaceholder.typicode.com/users`
  - `https://jsonplaceholder.typicode.com/posts`
- **(b)** Join the two datasets using `id` from users and `userId` from posts. Build a dictionary structured as: `{ user_name: { "email": "...", "post_count": N, "titles": [ list of post titles ] } }`. Print the result for the first 3 users only.
- **(c)** From the joined dictionary, find the user who has written the most posts. Print their: (i) full name, (ii) email address, (iii) company name, and (iv) all post titles numbered.
- **(d)** Filter users where either of the following is true: (i) `company.bs` field contains the word "synergies" (case-insensitive), or (ii) `address.city` has more than 8 characters. Print the name, city, and post count for all matching users.


```python
import requests
import pprint

# (a) Send GET requests and verify HTTP status code == 200
users_res = requests.get("https://jsonplaceholder.typicode.com/users", timeout=10)
if users_res.status_code != 200:
    raise ConnectionError(f"Users endpoint returned status code: {users_res.status_code}")
users_data = users_res.json()

posts_res = requests.get("https://jsonplaceholder.typicode.com/posts", timeout=10)
if posts_res.status_code != 200:
    raise ConnectionError(f"Posts endpoint returned status code: {posts_res.status_code}")
posts_data = posts_res.json()

print(f"(a) Successfully fetched {len(users_data)} users and {len(posts_data)} posts.")

# (b) Join datasets using id from users and userId from posts
joined_data = {}
for u in users_data:
    u_id = u["id"]
    u_name = u["name"]
    user_titles = [p["title"] for p in posts_data if p["userId"] == u_id]
    joined_data[u_name] = {
        "email": u["email"],
        "post_count": len(user_titles),
        "titles": user_titles
    }

print("\n(b) Joined Data for first 3 users:")
first_3_users = {k: joined_data[k] for k in list(joined_data.keys())[:3]}
pprint.pprint(first_3_users)

# (c) User who has written the most posts
max_user_name = max(joined_data.keys(), key=lambda name: joined_data[name]["post_count"])
max_user_obj = next(u for u in users_data if u["name"] == max_user_name)

print(f"\n(c) User with the most posts:")
print(f"    (i) Full Name   : {max_user_obj['name']}")
print(f"    (ii) Email      : {max_user_obj['email']}")
print(f"    (iii) Company   : {max_user_obj['company']['name']}")
print(f"    (iv) All Post Titles ({joined_data[max_user_name]['post_count']} posts):")
for idx, title in enumerate(joined_data[max_user_name]["titles"], start=1):
    print(f"        {idx:2d}. {title}")

# (d) Filter users: company.bs contains 'synergies' OR city > 8 characters
print(f"\n(d) Filtered Users (company.bs contains 'synergies' OR city > 8 chars):")
print(f"{'Name':<25} | {'City':<20} | {'Post Count'}")
print("-" * 55)
for u in users_data:
    bs_match = "synergies" in u["company"]["bs"].lower()
    city_match = len(u["address"]["city"]) > 8
    if bs_match or city_match:
        p_count = joined_data[u["name"]]["post_count"]
        print(f"{u['name']:<25} | {u['address']['city']:<20} | {p_count}")
```

## Task 8: To-Do Completion Rate Analysis Across Users

- **(a)** Fetch data from `https://jsonplaceholder.typicode.com/todos` and parse it into a Python list. Validate that: (i) the response status is 200, (ii) the parsed result is a list, and (iii) the list contains at least one record. If any validation fails, print a meaningful error and exit. Print the total number of todo records retrieved.
- **(b)** Group todos by `userId`. For each user, compute and store: `{ userId: { "total": N, "completed": N, "incomplete": N } }`. Print the result for all 10 users. For each user, verify that `completed + incomplete == total` and print True or False accordingly.
- **(c)** Compute the completion rate for each user as `completed / total × 100` rounded to 1 decimal place. Print: (i) the `userId` with the highest completion rate and its value, (ii) the `userId` with the lowest completion rate and its value, and (iii) the overall average completion rate across all users.
- **(d)** Fetch `https://jsonplaceholder.typicode.com/users` and build a lookup dictionary `{ userId: user_name }`. Print a formatted table with columns: `Rank | Name | Total Tasks | Completed | Completion Rate %`, sorted by completion rate in descending order. Use f-strings or `str.ljust()` to align columns neatly.


```python
import requests

# (a) Fetch todos, validate status == 200, is list, len > 0
todos_res = requests.get("https://jsonplaceholder.typicode.com/todos", timeout=10)
if todos_res.status_code != 200:
    print(f"Validation Error: Status code {todos_res.status_code}")
    exit(1)

todos_data = todos_res.json()
if not isinstance(todos_data, list) or len(todos_data) == 0:
    print("Validation Error: Response is not a non-empty list.")
    exit(1)

print(f"(a) Total todo records retrieved: {len(todos_data)}")

# (b) Group by userId: total, completed, incomplete. Verify completed + incomplete == total
user_todos = {}
for t in todos_data:
    uid = t["userId"]
    if uid not in user_todos:
        user_todos[uid] = {"total": 0, "completed": 0, "incomplete": 0}
    user_todos[uid]["total"] += 1
    if t["completed"]:
        user_todos[uid]["completed"] += 1
    else:
        user_todos[uid]["incomplete"] += 1

print("\n(b) To-Do Summary per User & Verification:")
for uid, stats in sorted(user_todos.items()):
    is_valid = (stats["completed"] + stats["incomplete"] == stats["total"])
    print(f"    User {uid:2d}: Total={stats['total']}, Completed={stats['completed']}, Incomplete={stats['incomplete']} | Verified: {is_valid}")

# (c) Completion rates per user, min, max, and overall average
rates = {uid: round((stats["completed"] / stats["total"]) * 100, 1) for uid, stats in user_todos.items()}
highest_u = max(rates.items(), key=lambda x: x[1])
lowest_u = min(rates.items(), key=lambda x: x[1])
overall_avg = round(sum(rates.values()) / len(rates), 1)

print(f"\n(c)(i) Highest completion rate: User {highest_u[0]} ({highest_u[1]}%)")
print(f"(c)(ii) Lowest completion rate: User {lowest_u[0]} ({lowest_u[1]}%)")
print(f"(c)(iii) Overall average completion rate: {overall_avg}%")

# (d) Fetch users, build lookup dictionary, and display formatted ranking table
users_res = requests.get("https://jsonplaceholder.typicode.com/users", timeout=10)
users_lookup = {u["id"]: u["name"] for u in users_res.json()}

sorted_ranks = sorted(rates.items(), key=lambda x: x[1], reverse=True)
header = f"{'Rank':<5} | {'Name':<25} | {'Total Tasks':<11} | {'Completed':<9} | {'Completion Rate %'}"
print(f"\n(d) Performance Ranking Table:\n{header}\n{'-' * len(header)}")
for rank, (uid, rate) in enumerate(sorted_ranks, start=1):
    name = users_lookup.get(uid, f"User {uid}")
    tot = user_todos[uid]["total"]
    comp = user_todos[uid]["completed"]
    print(f"{rank:<5} | {name:<25} | {tot:<11} | {comp:<9} | {rate}%")
```

## Task 9: Nested JSON Flattening, DataFrame Operations, and Comment Analysis

- **(a)** Fetch data from `https://jsonplaceholder.typicode.com/users`. Each user contains nested objects (`address` with `geo`, and `company`). Write a function `flatten_user(user: dict) -> dict` that returns a single-level dictionary with exactly these keys: `id`, `name`, `username`, `email`, `city`, `zipcode`, `lat`, `lng`, `company_name`, `company_catchphrase`. Apply this function to all 10 users and print the resulting flattened list.
- **(b)** Create a pandas DataFrame from the flattened user list and display it in full. Then: (i) cast the `lat` and `lng` columns to float, (ii) add a new column `distance_from_origin` computed as `sqrt(lat² + lng²)` using the `math` module applied row-by-row, and (iii) print the user with the greatest distance from origin.
- **(c)** Using the `lat` column, classify users into two groups: (i) Northern Hemisphere where `lat > 0`, and (ii) Southern Hemisphere where `lat < 0`. Print the names of users in each group. Then determine which hemisphere has the higher average `distance_from_origin`.
- **(d)** Fetch both endpoints below. For each post, count its number of comments by matching `postId` in comments against `id` in posts. Build a sorted list of tuples: `(post_id, post_title[:40], comment_count)` in descending order by comment count. Print the top 5 most-commented and bottom 5 least-commented posts.
  - `https://jsonplaceholder.typicode.com/posts`
  - `https://jsonplaceholder.typicode.com/comments`


```python
import math
import requests
import pandas as pd
from collections import Counter

# (a) Fetch users, flatten nested dictionary, apply to all 10 users
users_res = requests.get("https://jsonplaceholder.typicode.com/users", timeout=10)
users_raw = users_res.json()

def flatten_user(user: dict) -> dict:
    return {
        "id": user["id"],
        "name": user["name"],
        "username": user["username"],
        "email": user["email"],
        "city": user["address"]["city"],
        "zipcode": user["address"]["zipcode"],
        "lat": user["address"]["geo"]["lat"],
        "lng": user["address"]["geo"]["lng"],
        "company_name": user["company"]["name"],
        "company_catchphrase": user["company"]["catchPhrase"]
    }

flattened_users = [flatten_user(u) for u in users_raw]
print("(a) First 2 Flattened User Records:")
for u in flattened_users[:2]:
    print(" ", u)

# (b) Create DataFrame, cast lat/lng to float, compute distance_from_origin = sqrt(lat^2 + lng^2)
df_users = pd.DataFrame(flattened_users)
df_users["lat"] = df_users["lat"].astype(float)
df_users["lng"] = df_users["lng"].astype(float)
df_users["distance_from_origin"] = df_users.apply(
    lambda row: math.sqrt(row["lat"]**2 + row["lng"]**2), axis=1
)

print("\n(b) Full DataFrame with Distance from Origin:")
print(df_users[["id", "name", "city", "lat", "lng", "distance_from_origin"]].to_string(index=False))

farthest_row = df_users.loc[df_users["distance_from_origin"].idxmax()]
print(f"\n(b)(iii) User with greatest distance from origin:\n    {farthest_row['name']} (Distance: {farthest_row['distance_from_origin']:.4f})")

# (c) Hemisphere classification and comparison of average distances
north_df = df_users[df_users["lat"] > 0]
south_df = df_users[df_users["lat"] < 0]

print(f"\n(c)(i) Northern Hemisphere users ({len(north_df)}): {list(north_df['name'])}")
print(f"(c)(ii) Southern Hemisphere users ({len(south_df)}): {list(south_df['name'])}")

avg_north = north_df["distance_from_origin"].mean()
avg_south = south_df["distance_from_origin"].mean()
print(f"Average distance - North: {avg_north:.4f} | South: {avg_south:.4f}")
winner = "Northern Hemisphere" if avg_north > avg_south else "Southern Hemisphere"
print(f"Higher average distance: {winner}")

# (d) Fetch posts & comments, count comments per post, sort descending
posts_raw = requests.get("https://jsonplaceholder.typicode.com/posts", timeout=10).json()
comments_raw = requests.get("https://jsonplaceholder.typicode.com/comments", timeout=10).json()

comment_counts = Counter(c["postId"] for c in comments_raw)
post_comment_list = [
    (p["id"], p["title"][:40], comment_counts.get(p["id"], 0))
    for p in posts_raw
]
post_comment_list.sort(key=lambda x: x[2], reverse=True)

print("\n(d) Top 5 Most-Commented Posts:")
for pid, title, count in post_comment_list[:5]:
    print(f"    Post {pid:3d} | Comments: {count} | Title: {title}")

print("\n(d) Bottom 5 Least-Commented Posts:")
for pid, title, count in post_comment_list[-5:]:
    print(f"    Post {pid:3d} | Comments: {count} | Title: {title}")
```
